# Prepare the data for the K meanss clustering technique

```python
df.columns
```

```python
X = df[['Annual Income (k$)', 'Spending Score (1-100)']].values
print(f"features: Income and spending score")
print(X.shape)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
K = 5

km = KMeans(n_clusters=K, random_state = 42, n_init = 'auto')
km.fit(X_scaled)
```

```python
km.labels_
```

```python
print(silhouette_score(X_scaled, km.labels_))
```

# Run this code in a loop

```python
K_range = range(2, 11)
silhouette_scores = []

for k in K_range:
  km = KMeans(n_clusters=k, random_state = 42, n_init='auto')
  km.fit(X_scaled)
  silhouette_scores.append(silhouette_score(X_scaled, km.labels_))
```

```python
plt.plot(K_range, silhouette_scores, marker = 'o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score vs Number of Clusters')
plt.show()
```

# Fit the final model with the best K value

```python
kmeans = KMeans(n_clusters=5, random_state=42, n_init='auto')
labels = kmeans.fit_predict(X_scaled)

df['Cluster'] = labels
df
```

```python
# Scatter plot based on the clusters

sns.scatterplot(x = 'Annual Income (k$)',
                y = 'Spending Score (1-100)',
                hue = 'Cluster',
                data = df)

plt.show()
```
