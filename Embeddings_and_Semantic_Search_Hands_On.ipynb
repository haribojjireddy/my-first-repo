# Cosine Similary

```python
def cosine_similarity(vec1, vec2):
  vec1, vec2 = np.array(vec1), np.array(vec2)
  return float(np.dot(vec1, vec2)) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))
```

```python
def get_embeddings_hf(text):
  embeddings = model.encode(text)
  return embeddings

get_embeddings_hf("I love dogs")
```

```python
pairs = [
    ("I love dogs", "I love puppies"),
    ("I love dogs", "I love dogs"),
    ("I love dogs", "I love cats"),
    ("I love dogs", "The stock market has crashed"),
    ("I love dogs", "Dogs are wonderful pets"),
    ("Machine learning is cool", "Ml is awesome"),
    ("Machine Learnign is cool", "I ate pizza for lunch and dinner"),
    ("I am happy today", "I am very sad today")
]


print("=" * 65)
print("Semantic Similarity Test")
print("=" * 65)

for text1, text2 in pairs:
  emb1 = get_embedding(text1)
  emb2 = get_embedding(text2)
  sim = cosine_similarity(emb1, emb2)
  print(f"\n '{text1}' ")
  print(f" '{text2}' ")
  print(f" Similarity: {sim:.4f}")
```

```python
for text1, text2 in pairs:
  emb1 = get_embeddings_hf(text1)
  emb2 = get_embeddings_hf(text2)
  sim = cosine_similarity(emb1, emb2)
  print(f"\n '{text1}' ")
  print(f" '{text2}' ")
  print(f" Similarity: {sim:.4f}")
```

```python
# Does the model understand that same meaning in different languages should be similar?

test_pairs = [
    ("I am happy today", "నేను ఈ రోజు సంతోషంగా ఉన్నాను", "Same meaning (English/Telugu)"),
    ("I am happy today", "I am very sad today", "Opposite meaning (English/English)"),
    ("The food is delicious", "ఆహారం చాలా రుచిగా ఉంది", "Same meaning (food)"),
    ("The food is delicious", "The code has a bug", "Unrelated topics"),
    ("How are you?", "మీరు ఎలా ఉన్నారు?", "Same greeting"),
]

print("=" * 65)
print("🌐 CROSS-LANGUAGE SIMILARITY TEST")
print("=" * 65)

for text1, text2, label in test_pairs:
    emb1 = get_embedding(text1)
    emb2 = get_embedding(text2)
    sim = cosine_similarity(emb1, emb2)
    bar = '█' * int(sim * 30) + '░' * (30 - int(sim * 30))
    print(f"\n  [{label}]")
    print(f"    '{text1}'")
    print(f"    '{text2}'")
    print(f"    Similarity: {sim:.4f} [{bar}]")

print(f"\n{'─' * 65}")
print("💡 The embedding model understands meaning ACROSS languages!")
print("   Same meaning in Telugu and English → high similarity.")
print("   This is incredibly powerful for multilingual applications.")
```

# Build a semantic Serach oever docs

```python
# Our knowledge base — paragraphs about different ML topics
knowledge_base = [
    "Linear regression finds the best straight line through data points to predict continuous values like house prices or temperatures.",
    "Decision trees make predictions by asking a series of yes/no questions about the features, creating a flowchart-like structure.",
    "Random forests combine hundreds of decision trees and let them vote, reducing overfitting and improving accuracy.",
    "Overfitting happens when a model memorizes the training data including noise, performing great on training but poorly on new data.",
    "Cross-validation splits data into K folds and trains K times, each time using a different fold as the test set, giving a reliable performance estimate.",
    "Feature scaling normalizes numerical columns to the same range so that no single feature dominates the model due to its scale.",
    "Logistic regression uses the sigmoid function to predict probabilities for binary classification problems like spam detection.",
    "K-means clustering groups data points into K clusters by minimizing the distance between points and their cluster center.",
    "Gradient descent iteratively adjusts model parameters by moving in the direction that reduces the loss function.",
    "Regularization adds a penalty term to the loss function, discouraging overly complex models and preventing overfitting.",
    "The confusion matrix shows true positives, true negatives, false positives, and false negatives for classification models.",
    "Precision measures what fraction of positive predictions were actually correct, important when false positives are costly.",
    "Recall measures what fraction of actual positives were correctly identified, important when missing positives is costly.",
    "The ROC curve plots true positive rate vs false positive rate at different classification thresholds.",
    "SMOTE creates synthetic examples of the minority class to handle class imbalance in datasets.",
]


print(knowledge_base)
```

```python
kb_embeddings = get_embeddings(knowledge_base)
print(kb_embeddings)
```

```python
def semantic_search(query, top_k = 3):
  # Convert Wuery into Embeddings
  query_embedding = get_embedding(query)

  results = []

  for i, doc_emb in enumerate(kb_embeddings):
    sim = cosine_similarity(query_embedding, doc_emb) # The simialryt
    results.append((sim, i))


  # Sorting by similarity
  results.sort(reverse = True)

  # Display Results
  print(f"🔍 Query: '{query}' ")
  print(f"{'-'* 65}")
  for rank, (sim, idx) in enumerate(results[:top_k], 1):
    print(f"\n #{rank} [similarity: {sim:.4f}]")
    print(f"    {knowledge_base[idx]}")

  print()

  return results[:top_k]
```

```python
# Test with natural language queries (no keyword matching needed!)
print("=" * 65)
print("SEMANTIC SEARCH DEMO")
print("=" * 65)
print()

semantic_search("How do I stop my model from Memorizing the data?")
```
